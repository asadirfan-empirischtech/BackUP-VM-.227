```python
import torch
from diffusers import Cosmos3OmniPipeline

# Load the model using the specific Cosmos3OmniPipeline
pipe = Cosmos3OmniPipeline.from_pretrained(
    "nvidia/Cosmos3-Nano", 
    torch_dtype=torch.bfloat16, 
    device_map="cuda"
)

# Cosmos 3 expects prompts in a specific JSON structure for optimal quality
import json
json_prompt = {
    "type": "text2image",
    "prompt": "Astronaut in a jungle, cold color palette, muted colors, detailed, 8k"
}

# Run the inference
# For image generation, num_frames must be 1
result = pipe(
    prompt=json.dumps(json_prompt), 
    num_frames=1,
    height=720,
    width=1280
)

# Extract and save the image
image = result.images[0] # or result.video[0] depending on the exact output format of the pipeline
image.save("cosmos_test.jpg")
print("✅ Image successfully generated and saved!")
```

```python
import torch
import json
from diffusers import Cosmos3OmniPipeline

# Load the model with safety checker disabled
pipe = Cosmos3OmniPipeline.from_pretrained(
    "nvidia/Cosmos3-Nano", 
    torch_dtype=torch.bfloat16, 
    device_map="cuda",
    enable_safety_checker=False
)

# Detailed prompt focusing on your specific road obstacles and dashcam environment
json_prompt = {
    "type": "text2image",
    "prompt": "Dashcam view from a highway near majestic mountains, rough asphalt road surface with severe road deterioration, cracks, and potholes, scattered debris including loose rocks, fallen fabric bags, and broken car mirror shards on the road, overcast daylight, highly detailed, realistic driving perspective, 8k"
}

# Run the inference
result = pipe(
    prompt=json.dumps(json_prompt), 
    num_frames=1,
    height=720,
    width=1280
)

# Save the generated image
image = result.images[0]
image.save("mountain_highway_hazards.jpg")
print("✅ Road hazard image successfully generated and saved as 'mountain_highway_hazards.jpg'!")
```
